### Minard ###

```python
minard = Table.read_table('../../shared/Data/minard.csv')
minard
```

What percentage of Napoleon’s troops were still alive in each city of the advance and retreat?


```python
minard.select('City')
```

```python
minard.column('City')
```

```python
minard.column('Survivors')
```

```python
initial_size = minard.column('Survivors').item(0)
initial_size
```

```python
minard.column('Survivors') / initial_size
```

```python
minard = minard.with_column(
    'Pct alive', 
    minard.column('Survivors') / initial_size)
minard
```

```python
minard.set_format('Pct alive', PercentFormatter)
```

```python
minard.column("Pct alive")
```

```python
minard = minard.append_column("Pct alive 2", minard.column("Survivors") / minard.column("Survivors").item(0))
minard
```

```python
minard.set_format('Pct alive 2', PercentFormatter)
```
