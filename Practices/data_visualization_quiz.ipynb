### Question 1: Scatter Plots vs. Line Graphs

You have a dataset containing information about 1,000 top grossing movies from 1980 to 2015. 
You want to make two different plots:
1. A plot showing the **number of movies released each year** from 1980 to 2015.
2. A plot showing the **relationship between a movie's budget (numerical) and its box office gross (numerical)**.

**Questions:**
* **A)** For which plot should you use a **line graph (`.plot()`)**, and for which should you use a **scatter plot (`.scatter()`)**? Explain why based on the properties of the data.
* **B)** Suppose you try to use `.plot()` for the relationship between budget and gross. Describe what the resulting plot would look like and why it is not useful.

#### Solution to Question 1

<details>
<summary><b>Click to show Solution & Explanation</b></summary>

##### **Solution:**
* **Plot 1 (Movies per Year):** Use a **line graph (`.plot()`)**.
* **Plot 2 (Budget vs. Gross):** Use a **scatter plot (`.scatter()`)**.

##### **Explanation:**
* **Line Plots** are designed for **sequential/chronological data** (like time or distance). They are appropriate when:
  1. The $x$-axis has a natural order (years from 1980 to 2015).
  2. Sequential differences in $y$-values are meaningful.
  3. There is **only one $y$-value** (count of movies) for each $x$-value (year).
* **Scatter Plots** are designed for **non-sequential numerical variables** where you are looking for an **association**.
* If you try to use `.plot()` for Plot 2 (Budget vs. Gross), Python will draw lines connecting the points in the order they appear in the table. Because multiple movies can have the same or similar budgets but completely different grosses (multiple $y$-values for each $x$-value), the plot will look like an uninterpretable **"spaghetti" scribble** or a massive, tangled web of overlapping lines.

</details>

### Question 2: Bar Charts and Categorical Data

The method `.barh()` is used to plot categorical distributions.

**Questions:**
* **A)** Name three fundamental differences between a **Bar Chart (`.barh()`)** and a **Histogram (`.hist()`)**.
* **B)** Why is the order of the bars in a bar chart considered arbitrary, whereas the order of the bars in a histogram is fixed?

#### Solution to Question 2

<details>
<summary><b>Click to show Solution & Explanation</b></summary>

##### **Solution & Explanation:**
* **A) Key Differences:**
  1. **Data Types:** Bar charts display the distribution of **categorical variables** (e.g., movie genres, studios). Histograms display the distribution of **quantitative/numerical variables** (e.g., age, salary, years).
  2. **Spacing and Width:** In a bar chart, the width of the bars and the gap between them are arbitrary and chosen for visual styling. In a histogram, the bars are **contiguous** (they touch), and the width of each bar represents a numerical interval on a continuous number line.
  3. **Y-Axis Scale:** In a bar chart, bar length is proportional to the **count** or **percentage** in each category. In a true density histogram, the vertical axis is on the **density scale** (percent per unit), and the **area** of the bar represents the percentage.
* **B) Order of Bars:**
  * **Bar Chart:** The categories (e.g., "Action", "Comedy", "Drama") have no intrinsic numerical scale or chronological order. You can sort them alphabetically, by count, or any custom order.
  * **Histogram:** The horizontal axis is a continuous mathematical **number line**. The positions of the bins (e.g., $[10, 15)$, $[15, 20)$) are fixed mathematically and cannot be rearranged without destroying the meaning of the visualization.

</details>

### Question 3: The Area Principle & Density Histogram Math

You are given a density histogram of video game release years:
* The bin $[15, 20)$ (representing games released 15 to 20 years ago) has a height of **8% per year**.
* The bin $[20, 30)$ (representing games released 20 to 30 years ago) has a height of **3.8% per year**.

**Questions:**
* **A)** Calculate the percentage of video games that fall into the $[15, 20)$ bin.
* **B)** Calculate the percentage of video games that fall into the $[20, 30)$ bin.
* **C)** Why is the height of the $[20, 30)$ bin lower than the $[15, 20)$ bin, even though it covers a wider span of years? What does the height on a density scale represent?

```python
# Optional Scratchpad: Use Python to verify your calculations here
width_1 = 20 - 15
height_1 = 8
area_1 = width_1 * height_1

width_2 = 30 - 20
height_2 = 3.8
area_2 = width_2 * height_2

print("Bin [15, 20) percent:", area_1)
print("Bin [20, 30) percent:", area_2)
```

#### Solution to Question 3

<details>
<summary><b>Click to show Solution & Explanation</b></summary>

##### **Solution:**
* **A) Percentage in $[15, 20)$ bin:** **40%**
* **B) Percentage in $[20, 30)$ bin:** **38%**
* **C) Explanation:** The height represents **density** (crowdedness/concentration of data per unit of the horizontal axis). The $[15, 20)$ bin is more densely packed with games per year (8% per year over 5 years) than the $[20, 30)$ bin (3.8% per year over 10 years).

##### **Step-by-Step Math Walkthrough:**
* **For Bin $[15, 20)$:**
  * $\text{Width} = 20 - 15 = 5$ years.
  * $\text{Height} = 8\%$ per year.
  * $\text{Area (Percent)} = \text{Height} \times \text{Width} = 8\% \times 5 = 40\%$.
* **For Bin $[20, 30)$:**
  * $\text{Width} = 30 - 20 = 10$ years.
  * $\text{Height} = 3.8\%$ per year.
  * $\text{Area (Percent)} = \text{Height} \times \text{Width} = 3.8\% \times 10 = 38\%$.
* **Key Concept:** In a density histogram, you **must never read percentage directly from the vertical height**. The height is a rate: **percent per horizontal unit**. To get the actual percentage of the population in a bin, you must compute the area by multiplying height by width.

</details>

### Question 4: Unequal Bins

Suppose a variable is binned with unequal widths:
* Bin A: $[0, 5)$
* Bin B: $[5, 25)$

A researcher plots this data using `normed=False` (or `density=False`), which makes the height of each bar proportional to the raw count of entries inside it.

**Questions:**
* **A)** Why does this plot **violate the Area Principle** and misrepresent the distribution of the data?
* **B)** If Bin B contains 100 entries and Bin A contains 50 entries, explain how the count-based plot will look compared to a proper density-based histogram plot.

```python
# Optional Scratchpad: Density calculations for Question 4
count_A, width_A = 50, 5
count_B, width_B = 100, 20

density_A = count_A / width_A
density_B = count_B / width_B

print("Density of Bin A (entries/unit):", density_A)
print("Density of Bin B (entries/unit):", density_B)
```

#### Solution to Question 4

<details>
<summary><b>Click to show Solution & Explanation</b></summary>

##### **Solution & Explanation:**
* **A) Violation of the Area Principle:**
  * When bins are of unequal width, plotting raw counts or simple percentages on the vertical axis exaggerates the significance of wider bins.
  * The eye naturally perceives the **area** of the bar as its size/importance. Because Bin B is **4 times wider** than Bin A (width of 20 vs. 5), its bar's area will be artificially expanded, misleadingly suggesting there is a higher concentration of data there than there actually is.
* **B) Visual Comparison:**
  * **In the Count-Based Plot:** The bar for Bin B (100 count) will be **twice as tall** and **4 times wider** than the bar for Bin A (50 count). Its area will be $2 \times 4 = 8$ times larger, making it look massive and suggesting a highly concentrated region.
  * **In a True Density Histogram:** We divide by bin width to find density:
    * $\text{Density of Bin A} = \frac{50}{5} = 10$ entries per unit.
    * $\text{Density of Bin B} = \frac{100}{20} = 5$ entries per unit.
  * In the real histogram, **Bin A's bar is actually twice as tall as Bin B's bar**, showing that Bin A is twice as "crowded" or dense, even though Bin B has a higher total count because of its large width.

</details>

### Study Tip for Midterm 1:
If you are asked to find a missing height or percentage on the exam:
1. Calculate the **bin width** (Upper Limit - Lower Limit).
2. Look at the **height** on the vertical axis.
3. Multiply them to get the **percentage (area)**.
4. Remember that the **total area of all bars in a density histogram is always 100% (1.0)**. Use this fact if one of the bars is missing or cut off!
