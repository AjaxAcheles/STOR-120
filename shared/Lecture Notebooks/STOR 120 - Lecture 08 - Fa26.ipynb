## How many mentions?

```python
Twitter = Table.read_table('/home/jovyan/shared/Data/Twitter.csv')
Twitter
```

```python
'mississippi'.count('i')
```

```python
Twitter.column('Text').item(1)
```

```python
Twitter.column('Text').item(1).count('@')
```

```python
Twitter.column('Text').item(9)
```

```python
Twitter.column('Text').item(9).count('@')
```

```python
def Mention_Count(text):
    return text.count('@')
```

```python
Mention_Count(Twitter.column('Text').item(9))
```

```python
Mention_Count(Twitter.column('Text'))
```

## Apply ##

```python
Twitter.apply(Mention_Count, 'Text')
```

```python
Twitter_with_Mentions = Twitter.with_column(
    'Mentions',
    Twitter.apply(Mention_Count, 'Text')
)

Twitter_with_Mentions
```

```python
max(Twitter_with_Mentions.column('Mentions'))
```
