## Classifying Patients CKD ##

```python
ckd = Table.read_table('/home/jovyan/shared/Data/ckd.csv').relabeled('Blood Glucose Random', 'Glucose')
ckd.show(3)
```

```python
ckd.group('Class')
```

```python
ckd.scatter('White Blood Cell Count', 'Glucose', group = 'Class')
```

```python
ckd.scatter('Hemoglobin', 'Glucose', group = 'Class')
```

```python
# we want to be able to way to predict the class of someone
# one way to do this is to put some thresholds into code

max_glucose_for_0 = ckd.where('Class', 0).column('Glucose').max()
min_hemoglobin_for_0 = ckd.where('Class', 0).column('Hemoglobin').min()

max_glucose_for_0, min_hemoglobin_for_0
```

```python
def classify(hemoglobin, glucose):
    if hemoglobin < min_hemoglobin_for_0 or glucose > max_glucose_for_0:
        return 1
    else:
        return 0
```

```python
# Let's try our classifier!
classify(15, 100)
```

```python
ckd.scatter('Hemoglobin', 'Glucose', group = 'Class')
plots.scatter(15, 100, color='red');
```

```python
classify(10, 300)
```

```python
ckd.scatter('Hemoglobin', 'Glucose', group = 'Class')
plots.scatter(10, 300, color='red');
```

```python
np.mean(ckd.apply(classify, 'Hemoglobin', 'Glucose') == ckd.column('Class'))
```

```python
x_array = make_array()
y_array = make_array()

for x in np.arange(2.5, 18, 0.2):
    for y in np.arange(50, 500, 2):
        x_array = np.append(x_array, x)
        y_array = np.append(y_array, y)

test_grid = Table().with_columns(
    'Hemoglobin', x_array,
    'Glucose', y_array
)

test_grid.scatter('Hemoglobin')
```
