## Table Review: Sleep Study

```python
sleep = Table.read_table('/home/jovyan/shared/Data/SleepStudy.csv')
sleep.show(5)
```

```python
sleep.where('ClassYear', 4)
```

```python
sleep.where('ClassYear', are.equal_to(4))
```

```python
# Create a table with students who average more than 7, but less than 9 hours each night.
# Below is not quite what we want

sleep.where('AverageSleep', are.between(7,9))
```

```python
sleep.where('AverageSleep', are.strictly_between(7,9))
```

```python
# On average, how long do "Larks" sleep?
Larks = sleep.where('LarkOwl', 'Lark')
np.average(Larks.column('AverageSleep'))
```

```python
# How many students get at least 8 hours of sleep each night (on average)?
sleep.where('AverageSleep', are.above_or_equal_to(8)).num_rows
```

```python
# What proportion of students missed fewer than 3 classes?
sleep.where('ClassesMissed', are.below(3)).num_rows / sleep.num_rows
```

## What proportion of each age group were female in 2020?

```python
females_uspop_2020 = no_999.where('SEX', are.equal_to(2)).drop('SEX')
females_uspop_2020
```

```python
uspop_2020 = no_999.where('SEX', are.equal_to(0)).drop('SEX')
uspop_2020
```

```python
prop_female_uspop_2020 =  Table().with_column('Proportion Female', 
                                            females_uspop_2020.column('2020') / 
                                              uspop_2020.column('2020'))
prop_female_uspop_2020
```

```python
prop_female_uspop_2020 = prop_female_uspop_2020.with_column('AGE', females_uspop_2020.column('AGE'))
prop_female_uspop_2020
```

```python
prop_female_uspop_2020.plot('AGE', 'Proportion Female')
```
