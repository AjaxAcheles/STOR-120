```python
insurance = Table.read_table('/home/jovyan/shared/Data/insurance_charges.csv')
insurance
```

```python
insurance_smoker = insurance.where('smoker', 'yes')
insurance_smoker.scatter('bmi', 'charges')
```

### For a patient with a BMI of 26, how much will the charges be?

```python
insurance_smoker.scatter('bmi', 'charges')
plots.axvline(x=26, color='red', lw=2);
```

```python
insurance_smoker.where('bmi', are.equal_to(26))
```

### Can we use all BMIs *close* to 26? 

```python
insurance_smoker.scatter('bmi', 'charges')
plots.axvline(x=25, color='gold', lw=2);
plots.axvline(x=27, color='gold', lw=2);
```

```python
nearby = insurance_smoker.where('bmi', are.between(25, 27))
nearby_mean = np.mean(nearby.column('charges'))
nearby_mean
```

```python
insurance_smoker.scatter('bmi', 'charges')
plots.axvline(x=25, color='gold', lw=2);
plots.axvline(x=27, color='gold', lw=2);
plots.scatter(26, nearby_mean, color='red', s=50);
```

### Can we write a function to do this for all BMIs in the data? 

```python
def predict(bmi):
    nearby = insurance_smoker.where('bmi', are.between(bmi -1 , bmi + 1))
    return np.mean(nearby.column('charges'))
```

```python
predict(26)
```

```python
predict(23)
```

```python
predict(32)
```

```python
predicted_charges = insurance_smoker.apply(predict, 'bmi')
predicted_charges
```

```python
insurance_smoker = insurance_smoker.with_column(
    'Predicted charges', predicted_charges)
```

```python
insurance_smoker.select(
    'charges', 
    'bmi', 
    'Predicted charges'
).scatter('bmi')
```
