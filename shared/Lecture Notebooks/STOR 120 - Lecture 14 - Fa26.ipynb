### Empirical Distributions of a Statistic

```python
def sample_median(size):
    return np.median(music.sample(size).column('duration'))
```

```python
sample_median(10)
```

```python
num_simulations = 5000
```

```python
sample_medians_array = make_array()

for i in np.arange(num_simulations):
    new_median = sample_median(10)
    sample_medians_array = np.append(sample_medians_array, new_median)
```

```python
Table().with_column(
    'Sample medians', sample_medians_array
).hist(bins = np.arange(120,350, 10))
```

#### Empirical Distributions Overlayed

```python
sample_medians_array_10 = make_array()
sample_medians_array_100 = make_array()
sample_medians_array_1000 = make_array()

num_simulations = 5000

for i in np.arange(num_simulations):
    sample_medians_array_10 = np.append(
        sample_medians_array_10, 
        sample_median(10)
    )
    
    sample_medians_array_100 = np.append(
        sample_medians_array_100, 
        sample_median(100)
    )
    
    sample_medians_array_1000 = np.append(
        sample_medians_array_1000, 
        sample_median(1000)
    )
```

```python
sample_medians = Table().with_columns('Size 10', sample_medians_array_10, 
                                      'Size 100', sample_medians_array_100,
                                      'Size 1000', sample_medians_array_1000)
```

```python
sample_medians.hist(bins = np.arange(170,300, 1))
```

## Instagram Posts ##

```python
bieber = Table.read_table('/home/jovyan/shared/Data/bieber.csv')
bieber
```

```python
# 1609477260 1/1/2021 midnight
# 1617249600 4/1/2021 midnight

bieber_2021Q1 = bieber.where('post_dates', are.between(1609477260, 1617249600))
bieber_2021Q1                           
```

```python
observed_video_prop = sum(bieber_2021Q1.column('is_video')) / bieber_2021Q1.num_rows
observed_video_prop
```

```python
sample_size = bieber_2021Q1.num_rows
sample_size
```

```python
predicted_proportions = make_array(0.166, 1 - 0.166)
sample_proportions(sample_size, predicted_proportions)
```

```python
def simulated_bieber_post():
    return sample_proportions(sample_size, predicted_proportions).item(0)
```

```python
simulated_bieber_post()
```

```python
bieber_simulated_array = make_array()

for i in np.arange(10000):
    bieber_simulated_array = np.append(bieber_simulated_array , simulated_bieber_post())
```

```python
Table().with_column('Proportion of Bieber Instagram Posts with Video', bieber_simulated_array).hist()
plots.axvline(observed_video_prop, color='red', lw=2);
```

```python
np.mean(bieber_simulated_array >= observed_video_prop)
```

## Swain vs. Alabama ##

```python
population_proportions = make_array(.26, .74)
population_proportions
```

```python
sample_proportions(100, population_proportions)
```

```python
def panel_proportion():
    return sample_proportions(100, population_proportions).item(0)
```

```python
panel_proportion()
```

```python
panels = make_array()

num_simulations = 100000

for i in np.arange(num_simulations):
    new_panel = panel_proportion() * 100
    panels = np.append(panels, new_panel)
```

```python
Table().with_column('Number of Black Men on Panel of 100', panels).hist(bins=np.arange(5.5,40.))
plots.axvline(8, color='red', lw=2);
```

```python
np.mean(panels <= 8)
```
